```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
salaries = pd.read_csv("../data/cleaned/salaries_clean.csv")
salaries
```

```python
exp_salary = salaries.groupby('experience_level')['salary_in_usd'].mean().sort_values(ascending=False)
plt.bar(exp_salary.index, exp_salary.values, color="orange")
plt.xlabel("Experience Level")
plt.ylabel("Average Salary (USD)")
plt.title("Average Salary by Experience Level")
plt.grid()
plt.show()
```

```python
year_salary = salaries.groupby('work_year')['salary_in_usd'].mean().sort_index()
plt.bar(year_salary.index, year_salary.values, color="navy")
plt.xlabel("Year")
plt.ylabel("Average Salary (USD)")
plt.title("Average Salary Trend by Year")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='remote_ratio', y='salary_in_usd', data=salaries, color="y")
plt.xlabel("Remote Ratio (0=No Remote, 50=Hybrid, 100=Full Remote)")
plt.ylabel("Salary (USD)")
plt.title("Salary Distribution by Remote Work Type")
plt.show()
```

```python
top_titles = salaries['job_title'].value_counts().head(10)

plt.figure(figsize=(8,5))
plt.bar(top_titles.index, top_titles.values, color="teal")
plt.xlabel("Job Title")
plt.ylabel("Number of Postings")
plt.title("Top 10 Job Titles")
plt.xticks(rotation=45, ha='right') # using for job titltle name curved
plt.show()
```

```python
numeric_cols = salaries[['salary_in_usd', 'remote_ratio', 'work_year']]
correlation = numeric_cols.corr()

plt.figure(figsize=(6,4))
sns.heatmap(correlation, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```python
#One line purpose for each chart:

#Chart	Purpose
#Chart 1 (Experience)	Who earns more — Entry level or Executive level
#Chart 2 (Year)	Did average salary go up or down over the years
#Chart 3 (Remote)	Do remote jobs pay more than office jobs
#Chart 4 (Job Titles)	Which job title appears most often in the dataset
#Correlation	Are salary, remote work, and year connected to each other
```
