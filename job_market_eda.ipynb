```python
# Job Market Abalysis
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
salaries = pd.read_csv("../data/raw/ds_salaries.csv")
salaries
salaries.head()
```

```python
# chek shape

salaries.shape
```

```python
# check info
salaries.info()
```

```python
salaries.dtypes
```

```python
# check misiing values
salaries.isnull().sum()
```

```python
# Confirmed above: no missing values in this dataset, so no fillna needed
# (kept as a comment for reference — nothing to fill)
```

```python
# drop column unnamed
salaries = salaries.drop(columns=["Unnamed: 0"])
salaries.head()
```

```python
# checing duplicates values

print(salaries.duplicated().sum())
```

```python
# we have 42 duplicate rows — remove them
salaries = salaries.drop_duplicates()
```

```python
salaries.shape
```

```python
salaries.dtypes
```

```python
# suppose object in int so convert into text   we have accurate dtypes 
salaries['company_size'] = salaries['company_size'].astype(str)
```

```python
# confirm chek
print(salaries['company_size'].dtype)
```

```python
# we can be see which types data stored in column 
print(salaries["company_size"].unique())
```

```python
# save data
salaries.to_csv("../data/cleaned/salaries_clean.csv", index=False)
print("saved")
```
